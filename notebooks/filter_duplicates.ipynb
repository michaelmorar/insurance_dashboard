# Load Aviva Resource Tracker

This notebook locates the Excel file in the `inputs` folder (either `resource.xlsx` or `Aviva Resource Tracker_May 2025.xlsm`) using absolute paths to prevent any working directory resolution errors, and loads the data into a pandas DataFrame.

```python
import os
import pandas as pd
import warnings

# Suppress openpyxl conditional formatting warnings
warnings.filterwarnings('ignore', category=UserWarning, module='openpyxl')
```

```python
def get_excel_path(project_root, spreadsheet):
    possible_paths = [
        os.path.join(project_root, 'inputs', spreadsheet)
    ]   
    print(possible_paths)
    excel_path = None
    for path in possible_paths:
        if os.path.exists(path):
            excel_path = path
            break

    if excel_path is None:
        raise FileNotFoundError(
            f"Could not find any resource tracker Excel file in the inputs directory.\n"
            f"Looked for:\n" + "\n".join(possible_paths)
        )

    return excel_path
```

```python
# Define the absolute project root directory
project_root = "/Users/MMIE/projects/insurance_dashboard"

# Check if the current working directory is stale and recover it if needed
try:
    cwd = os.getcwd()
    print(f"Current working directory: {cwd}")
except FileNotFoundError:
    print("WARNING: The Jupyter kernel's working directory is stale (likely because a directory was deleted/recreated).")
    if os.path.exists(project_root):
        os.chdir(project_root)
        print(f"Successfully recovered working directory by switching to project root: {os.getcwd()}")
    else:
        print("Could not find the project root directory. Please check the project path.")

# Define potential absolute paths to the Excel files

resource_sheet = get_excel_path(project_root, "resource.xlsx")

print(f"Using Excel file: {resource_sheet}")

# Load resource_sheet to a dataframe
df = pd.read_excel(resource_sheet)
```

```python
# Load Resource sheet to Dataframe
resource_dataframe = pd.read_excel(resource_sheet, sheet_name="Sheet1")

print(f"Successfully loaded data.")
print(f"DataFrame shape: {resource_dataframe.shape[0]} rows, {resource_dataframe.shape[1]} columns.")
```

```python

# Selected candidate names where date rolled off is 2026 and status is "rolled off"
rolled_off_candidates = resource_dataframe[
    (pd.to_datetime(resource_dataframe['Resource Roll Off Date'], errors='coerce').dt.year == 2026) &
    (resource_dataframe['Status'].str.contains('rolled off', case=False, na=False))
]['Candidate Name']

print(rolled_off_candidates)
print(len(rolled_off_candidates))
```

```python
# Load Resource sheet to Dataframe
dashboard_sheet = get_excel_path(project_root, "dashboard.xlsx")

print(f"Using Excel file: {dashboard_sheet}")

# Load resource_sheet to a dataframe
df = pd.read_excel(dashboard_sheet)


dashboard_dataframe = pd.read_excel(dashboard_sheet, sheet_name="Sheet1")

print(f"Successfully loaded data.")
print(f"DataFrame shape: {dashboard_dataframe.shape[0]} rows, {dashboard_dataframe.shape[1]} columns.")
```
